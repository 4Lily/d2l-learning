# 5.5. 读写文件

本节学习如何把张量与模型参数保存到磁盘，并重新加载。请自己完成每个代码单元。

导入所需模块：torch、torch.nn、torch.nn.functional（F）。

```python
import torch
from torch import nn
import torch.nn.functional as F
```

### 5.5.1. 加载和保存张量

保存单个张量：
- x = torch.arange(4)；torch.save(x, 'x-file')
- 读取：x2 = torch.load('x-file')；x2

```python
x=torch.arange(4)
torch.save(x,'x-file')
x2=torch.load('x-file')
x2
```

保存张量列表：
- y = torch.zeros(4)；torch.save([x, y], 'x-files')
- x2, y2 = torch.load('x-files')；(x2, y2)

```python
y = torch.zeros(4)
torch.save([x, y], 'x-files')
x2, y2 = torch.load('x-files')
(x2, y2)
```

保存字符串到张量的字典：
- mydict = {'x': x, 'y': y}；torch.save(mydict, 'mydict')
- mydict2 = torch.load('mydict')；mydict2

```python
mydict={'x':x,'y':y}
mydict=torch.save(mydict,'mydict')
mydict2=torch.load('mydict')
```

### 5.5.2. 加载和保存模型参数

先定义一个多层感知机 MLP 类并实例化、前向：
- class MLP(nn.Module)：
  - __init__：super().__init__()；self.hidden = nn.Linear(20, 256)；self.output = nn.Linear(256, 10)
  - def forward(self, x): return self.output(F.relu(self.hidden(x)))
- net = MLP()；X = torch.randn(size=(2, 20))；Y = net(X)

```python
from turtle import forward


class MLP(nn.Module):
    def __init__(self) -> None:
        super().__init__()
        self.hidden=nn.Linear(20,256)
        self.output=nn.Linear(256,10)
    def forward(self,X):
        return self.output(F.relu(self.hidden(X)))
net=MLP()
X=torch.randn(size=(2,20))
Y=net(X)
```

保存模型参数：
- torch.save(net.state_dict(), 'mlp.params')

重新实例化并加载参数，验证一致：
- clone = MLP()；clone.load_state_dict(torch.load('mlp.params'))；clone.eval()
- Y_clone = clone(X)；Y_clone == Y

```python
torch.save(net.state_dict(),'mlp.params')
clone = MLP()
clone.load_state_dict(torch.load('mlp.params'))
clone.eval()
Y_clone=clone(X)
Y_clone==Y
```

### 5.5.3. 小结

- save / load 函数可用于张量对象的文件读写
- 可以通过参数字典保存和加载网络的全部参数
- 保存架构必须在代码中完成，而不是在参数中
