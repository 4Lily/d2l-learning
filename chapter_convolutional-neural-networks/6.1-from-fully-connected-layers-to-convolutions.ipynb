# 6.1. 从全连接层到卷积

图像具有二维空间结构。将图像展平后输入全连接层会丢失像素的邻近关系，并产生大量参数。卷积神经网络利用图像的平移不变性和局部性来解决这一问题。

### 6.1.1 不变性

同一个目标出现在图像的不同位置时，识别结果应保持不变，这体现了平移不变性。图像局部区域中的像素通常具有更强的相关性，因此模型可以只连接局部感受野。

### 6.1.2 多层感知机的限制

全连接层对每个输入位置使用独立参数，无法自然利用平移结构；当图像分辨率增大时，参数量会快速增长。

### 6.1.3 卷积

卷积层在局部窗口内共享同一组权重。二维互相关的输出尺寸为 `(n_h-k_h+1) × (n_w-k_w+1)`；加入填充和步幅后可进一步控制尺寸。

请思考：权重共享如何带来平移不变性？局部连接如何减少参数量？

### 6.1.4 “沃尔多在哪里”回顾

检测图像中的局部模式时，同一个模式可能出现在任意位置。卷积核可以在整幅图像上滑动，从而在不同位置寻找相同特征。

### 6.1.5 小结与练习

- 卷积神经网络通过局部性和权重共享处理图像。
- 平移不变性使同一特征可以在不同位置被识别。
- 逐层堆叠卷积可以扩大感受野并组合更复杂的特征。

练习：比较全连接层和卷积层的参数量；说明为什么卷积层更适合高分辨率图像。
