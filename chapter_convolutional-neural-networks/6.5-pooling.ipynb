# 6.5. 汇聚层

汇聚层在局部窗口内进行最大值或平均值聚合，降低空间分辨率并提供一定的平移鲁棒性。它通常没有可学习参数。

### 6.5.1 最大汇聚层和平均汇聚层

先实现 `pool2d(X, pool_size, mode)`，支持 `max` 和 `avg`。在 3×3 输入上测试 2×2 窗口，并解释最大汇聚为什么更关注显著特征。

### 6.5.2 填充和步幅

用 `nn.MaxPool2d` 测试默认步幅、显式填充和步幅为 2 的情况。比较不同设置下输出张量的形状。

### 6.5.3 多个通道

汇聚层对每个通道独立操作，因此输入通道数和输出通道数相同。构造两个通道的输入，验证通道之间不会互相混合。

### 6.5.4 小结与练习

练习：比较最大汇聚和平均汇聚的输出；修改窗口和步幅；说明汇聚层如何减少后续层的计算量。
