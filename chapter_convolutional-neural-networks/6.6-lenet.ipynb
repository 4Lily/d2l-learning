# 6.6. 卷积神经网络（LeNet）

LeNet 是早期成功的卷积神经网络，由卷积层、汇聚层和全连接层组成，可用于 Fashion-MNIST 图像分类。

### 6.6.1 LeNet

使用 `nn.Sequential` 构建 LeNet：

1. `Conv2d(1, 6, kernel_size=5)`、`Sigmoid`、`MaxPool2d(2, 2)`；
2. `Conv2d(6, 16, kernel_size=5)`、`Sigmoid`、`MaxPool2d(2, 2)`；
3. 展平后接 `Linear(16*4*4, 120)`、`Sigmoid`、`Linear(120, 84)`、`Sigmoid`、`Linear(84, 10)`。

用随机输入 `(1, 1, 28, 28)` 检查各层输出形状。

### 6.6.2 模型训练

导入 `torch`、`torch.nn` 和 `d2l.torch`，加载 Fashion-MNIST。设置 `batch_size`、学习率和训练轮数，使用交叉熵损失与 SGD 优化器训练 LeNet。实现或调用 GPU 上的准确率评估函数，并绘制训练损失、训练准确率和测试准确率。

### 6.6.3 小结

- LeNet 通过卷积层提取局部特征，通过汇聚层降低空间分辨率，再用全连接层完成分类。
- 卷积和汇聚共同利用了图像的空间结构。
- 现代网络通常使用 ReLU、批量归一化和更深的架构，但基本思想仍来自这类卷积网络。

### 6.6.4 练习

1. 将 LeNet 中的 Sigmoid 替换为 ReLU，比较收敛速度和准确率。
2. 调整卷积通道数、全连接层宽度或学习率。
3. 将最大汇聚替换为平均汇聚。
4. 在训练过程中记录每个 epoch 的耗时，并比较 CPU 与 GPU。
5. 可视化第一层卷积核或中间特征图。
