# 6.3. 填充和步幅

不加处理时卷积会缩小特征图。填充在边界外补零，步幅控制卷积窗口每次移动的距离。

### 6.3.1 填充

对于奇数大小的卷积核，通常在两侧填充 `k//2` 个零，使输出高宽与输入相同。使用 `nn.Conv2d` 验证 `kernel_size=(5,3), padding=(2,1)` 的输出尺寸。

### 6.3.2 步幅

步幅为 2 时窗口每次跨越两个像素，输出尺寸大约减半。分别测试整数步幅、二维步幅和非对称填充。输出尺寸公式为 `floor((n+2p-k)/s)+1`。

```python
import torch
from torch import nn
conv2d=nn.Conv2d(1,3,kernel_size=(5,3), padding=(2,1),stride=2)//输入通道，输出通道，卷积核尺寸，填充，步幅
x=torch.rand(1,3,3)
print(conv2d(x).shape)
```

### 6.3.3 小结与练习

- 填充控制边界信息和输出尺寸。
- 步幅控制窗口移动速度，也可以用于下采样。

练习：计算不同 `(padding, stride, kernel_size)` 组合的输出尺寸；验证偶数大小卷积核无法用对称填充完全保持尺寸。
